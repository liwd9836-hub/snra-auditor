# SNRA quickstart tutorial v1.0

This notebook is a lightweight submission-facing walkthrough. It uses the local package, toy data, and bundled manifest summaries. It does not require Docker or optional heavy spatial packages.

## 1. Install locally

From the repository root, install the package in editable mode:

```bash
python -m pip install -e .
```

````python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
print(ROOT)
````

## 2. Check the CLI

The installed console command is `snra`. The module form is useful before installation.

````python
subprocess.run([sys.executable, "-m", "src.snra.cli_v0_1", "--help"], cwd=ROOT, check=True)
````

## 3. Run the toy audit-card demo

This writes a compact summary TSV and an audit-card markdown file.

````python
demo_out = ROOT / "results" / "snra_runtime_memory_v1_0" / "notebook_demo"
subprocess.run(
    [sys.executable, "-m", "src.snra.cli_v0_1", "demo", "--out-dir", str(demo_out)],
    cwd=ROOT,
    check=True,
)
sorted(path.name for path in demo_out.iterdir())
````

## 4. Reproduce the bundled core manifest

Failed certificates in the bundled manifest are expected audit outcomes, not software failures.

````python
manifest_out = ROOT / "results" / "snra_runtime_memory_v1_0" / "notebook_core_manifest"
subprocess.run(
    [
        sys.executable,
        "-m",
        "src.snra.cli_v0_1",
        "manifest",
        "--manifest-json",
        str(ROOT / "metadata" / "snra" / "snra_manifest_core_evidence_v0_1.json"),
        "--out-dir",
        str(manifest_out),
    ],
    cwd=ROOT,
    check=True,
)
(manifest_out / "snra_manifest_audit_summary_v0_1.tsv").read_text(encoding="utf-8").splitlines()[:3]
````

## 5. Run lightweight runtime/memory profiling

This uses synthetic inputs and bundled core manifest summaries, then updates the runtime report.

````python
subprocess.run([sys.executable, "scripts_profile_snra_runtime_memory_v1_0.py"], cwd=ROOT, check=True)
````

## Evidence boundary

SNRA outputs are audit and claim-calibration artifacts. They support reproducible certificate, downgrade, or block decisions for submitted spatial claims, but they do not establish causal tumor-immune or tumor-stroma mechanisms.
